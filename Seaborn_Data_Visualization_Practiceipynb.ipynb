```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

##### import tips dataset.

```python
df=sns.load_dataset("tips")
df
```

##### 1. Write a Python program to display a bar chart of the total bill of Gender and write the intuition about the plot.

```python
plt.figure(figsize=(5,4))
sns.barplot(data=df, x="sex", y="total_bill")
plt.title("Average Total Bill by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Total Bill")
plt.show()
```

```python
### Intuition:

# *This bar chart shows the average total bill for males and females. Males have a slightly higher average total bill than females.
```

###### 2. Write a Python program to display a box plot of the total bill of day and hue = "smoker" and write the intuition about the plot.

```python
plt.figure(figsize=(5,4))
sns.boxplot(data=df, x="day", y="total_bill", hue="smoker")
plt.title("Total Bill by Day and Smoker")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
### Intuition:

# The box plot shows the distribution of total bills across different days for smokers and non-smokers.
# Smokers generally have higher and more variable total bills on some days, with a few high-value outliers.
```

- Smokers generally spend more on weekends, especially on Saturday and Sunday. Total bills are higher on weekends for both groups, with more outliers among smokers.

##### 3. Write a Python program to display a violin plot of the total bill of Day and hue = "sex" and write the intuition about the plot.

```python
plt.figure(figsize=(5,4))
sns.violinplot(data=df, x="day", y="total_bill", hue="sex")
plt.title("Total Bill by Day and Gender")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
### Intuition:

#The violin plot shows the distribution and density of total bills for males and females across different days.
#It helps compare where the bills are more concentrated and how widely they are distributed.
```

##### 4. Write a Python program to display a strip plot of the total bill of Day and hue = "sex" and write the intuition about the plot.

```python
plt.figure(figsize=(5,4))
sns.stripplot(data=df, x="day", y="total_bill", hue="sex")
plt.title("Total Bill by Day and Gender")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
### Intuition:

# The strip plot displays individual total bill observations for males and females across different days.
# Most bills are concentrated at lower values, while some customers have higher bills.
```

##### 5. Write a Python program to display heat map of dataframe and write the intuition about the plot.

```python
plt.figure(figsize=(8, 6))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

```python
### Intuition:

# The heatmap shows the correlation between numerical variables.
# Total bill and tip have a positive correlation, meaning customers tend to give higher tips when the total bill is higher.
# Size also has a positive relationship with total bill.
```

##### 6. Write a Python program to display scatter plot of total bill of tip and write the intuition about the plot.

```python
plt.figure(figsize=(5,4))
sns.scatterplot(data=df, x="total_bill", y="tip")
plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
### Intuition:

# The scatter plot shows a positive relationship between total bill and tip. As the total bill increases, the tip generally tends to increase.
```

##### 7. Write a Python program to display density plot of all numerical columns from the dataframe and write the intution about plot.

```python
plt.figure(figsize=(5,4))
num_cols = df.select_dtypes(include="number").columns

for col in num_cols:
    sns.kdeplot(data=df, x=col, fill=True)
    plt.title(f"Density Plot of {col}")
    plt.show()
```

```python
### Intuition:

# The density plot shows the distribution of all numerical variables.
# Most observations for tip and size are concentrated around smaller values, while total bills have a wider distribution with a longer tail toward higher values.
```

- This density plot shows that most tips and group sizes are small and tightly clustered, while total bills vary more widely with a longer tail toward higher amounts.

##### 8. Write a Python program to calculate average total bills by days and display top 4 days by plot.

```python
avg_bill = df.groupby("day")["total_bill"].mean().sort_values(ascending=False)

print(avg_bill)
```

```python
plt.figure(figsize=(5,4))
top4 = avg_bill.head(4)

sns.barplot(x=top4.index, y=top4.values)

plt.title("Top 4 Days by Average Total Bill")
plt.xlabel("Day")
plt.ylabel("Average Total Bill")
plt.show()
```

- This bar chart shows that Sunday has the highest average total bill, followed by Saturday, while Friday has the lowest among the top 4 days.

#### Bonus Question

###### 1. What is Histogram chart ?

A histogram is an accurate graphical representation of the distribution of a numeric variable. It takes as input a numeric variable only. The variable is cut into several bins, and the number of observation per bin is represented by the height of the bar.

##### 2. What is box plot ?

A box-plot is a very useful and standardized way of displaying the distribution of data based on a five-number summary (minimum, first quartile, second quartile(median), third quartile, maximum). It helps in understanding these parameters of the distribution of data and is extremely helpful in detecting outliers.

##### 3. What are the plots we can use for univeriate analysis.

- count plot
- bar plot
- box plot
- hist plot
- kde
- pie chart

##### 4. What are the plots we can use for bivariate analysis.

- scatter plot
- box plot
- grouped bar plot

###### 5. What are plot we can use for multivariate analysis.

- pair plot
- heat map

Histogram → Used to show distribution of numerical data.

Box Plot → Used to show distribution, median, quartiles and outliers.

Univariate Analysis → Count Plot → Bar Plot → Histogram → Box Plot → KDE → Pie Chart

Bivariate Analysis → Scatter Plot → Bar Plot → Box Plot → Line Plot → Violin Plot

Multivariate Analysis → Pair Plot → Heatmap → Scatter Plot with Hue → Grouped Bar Plot → Violin Plot with Hue

```python
#1. What is Histogram chart ?
sns.histplot(data=df, x="total_bill", bins=10)
plt.title("Distribution of Total Bill")
plt.show()
```

```python
#2. What is box plot ?
plt.figure(figsize=(5,4))
sns.boxplot(data=df, x="total_bill")
plt.title("Box Plot of Total Bill")
plt.show()
```

```python
#3. What are the plots we can use for univeriate analysis.
# Numerical variable:
plt.figure(figsize=(5,4))
sns.histplot(data=df, x="total_bill")
plt.show()
# Categorical variable:
plt.figure(figsize=(5,4))
sns.countplot(data=df, x="sex")
plt.show()
```

```python
# 4. What are the plots we can use for bivariate analysis.
## Numerical vs Numerical:
plt.figure(figsize=(5,4))
sns.scatterplot(data=df, x="total_bill", y="tip")
plt.show()
## Categorical vs Numerical:
plt.figure(figsize=(5,4))
sns.boxplot(data=df, x="day", y="total_bill")
plt.show()
```

```python
##5. What plots can we use for Multivariate Analysis?

## Example 1 – Pair Plot
sns.pairplot(df)
plt.show() #This compares multiple numerical variables with each other.

#Example 2 – Heatmap
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.show() #This shows correlations between multiple numerical variables.

##Example 3 – Scatter Plot with Hue
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="sex"
)

plt.show()
```
